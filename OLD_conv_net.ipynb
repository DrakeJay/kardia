```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import numpy as np
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
class ConvNet(nn.Module):
  def __init__(self, in_channels, out_channels):
    super().__init__()
    self.conv = nn.Sequential(
        nn.Conv2d(in_channels, out_channels, kernel_size= 3, stride= 1, padding= 1, bias=False),
        nn.BatchNorm2d(out_channels),
        nn.ReLU(inplace=True),
        nn.Conv2d(out_channels, out_channels,kernel_size= 3, stride= 1, padding= 1, biaS=False),
        nn.BatchNorm2d(out_channels),
        nn.Relu(inplace=True)
    )

  def forward(self, x):
    return self.conv(x)
```

##class ConvNet(nn.Module):
* This line declares a new class ConvNet that inherits from nn.Module, which is the base class for all neural network modules in PyTorch.

##__init__(self, in_channels, out_channels):
* This is the constructor for the ConvNet class. It initializes the layers of the network.

##super().__init__():
* Calls the constructor of the parent class nn.Module.

##self.conv = nn.Sequential(...):
* This defines a sequential container of layers, meaning the data will flow through these layers in the order they are defined.

##nn.Conv2d(in_channels, out_channels, kernel_size=3, stride=1, padding=1, bias=False):
* This is a 2D convolutional layer.
* It takes in_channels as input and produces out_channels.
* The kernel_size=3 means it uses a 3x3 filter,
* stride=1 means it moves the filter one pixel at a time,
* padding=1 adds one pixel padding around the input to maintain the spatial dimensions.
* bias=False indicates that a learnable bias term is not added to the output (often omitted when followed by BatchNorm).

##nn.BatchNorm2d(out_channels):
* This is a 2D Batch Normalization layer. It normalizes the outputs of the previous convolutional layer across mini-batches, which helps stabilize and speed up training.

##nn.ReLU(inplace=True):
* This is a Rectified Linear Unit activation function. It introduces non-linearity by outputting max(0, x).

##inplace=True
* modifies the input directly, potentially saving memory.

* The second set of nn.Conv2d, nn.BatchNorm2d, and nn.ReLU layers forms another identical block of operations, further processing the features

##forward(self, x):
* This method defines the forward pass of the network. It specifies how the input x flows through the layers defined in __init__.

##return self.conv(x):
* The input x is passed through the self.conv sequential module, and the output of this sequence is returned.

#Conv2d

  * in_channels (int): Number of channels in the input image

  * out_channels (int): Number of channels produced by the
    convolution

  * kernel_size (int or tuple): Size of the convolving kernel

  * stride (int or tuple, optional): Stride of the convolution,
    Default: 1

  * padding (int, tuple or str, optional): Padding added to all four sides ofthe input. Default: 0

  * dilation (int or tuple, optional): Spacing between kernel elements. Default: 1

  * groups (int, optional): Number of blocked connections from input
      channels to output channels. Default: 1

  * bias (bool, optional): If ``True``, adds a learnable bias to the
      output. Default: ``True``

  * padding_mode (str, optional): ``'zeros'``, ``'reflect'``,
      ``'replicate'`` or ``'circular'``. Default: ``'zeros'``
